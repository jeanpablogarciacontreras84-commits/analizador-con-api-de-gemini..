```python
# ==============================================================================
# 🌟 TERMINAL DE INTELIGENCIA FINANCIERA & TRADING - UI MODERNA & 30 REINTENTOS
# ==============================================================================

# 1. INSTALACIÓN DE DEPENDENCIAS
!pip install -q feedparser yfinance google-genai pandas numpy ipywidgets

# 2. IMPORTACIÓN DE LIBRERÍAS
import time
import feedparser
import yfinance as yf
import pandas as pd
import numpy as np
from google import genai
from google.genai import types
import urllib.parse
from IPython.display import Markdown, display, clear_output, HTML
import ipywidgets as widgets

# 3. DISEÑO CSS MODERNO Y ANIMACIONES FLUIDAS
css_estilos = HTML("""
<style>
@import url('https://fonts.googleapis.com/css2?family=Inter:wght@300;400;500;600;700&display=swap');

.fintech-container {
    font-family: 'Inter', sans-serif;
    background: linear-gradient(135deg, #0f172a 0%, #1e293b 100%);
    border-radius: 16px;
    padding: 24px;
    box-shadow: 0 10px 25px -5px rgba(0, 0, 0, 0.4), 0 8px 10px -6px rgba(0, 0, 0, 0.4);
    color: #f8fafc;
    border: 1px solid rgba(255, 255, 255, 0.1);
}

.fintech-header {
    font-size: 22px;
    font-weight: 700;
    color: #38bdf8;
    margin-bottom: 16px;
    display: flex;
    align-items: center;
    gap: 10px;
    border-bottom: 2px solid rgba(56, 189, 248, 0.2);
    padding-bottom: 10px;
}

/* Estilo personalizado para campos de entrada */
input, select, textarea {
    background-color: #1e293b !important;
    color: #f8fafc !important;
    border: 1px solid #334155 !important;
    border-radius: 8px !important;
    padding: 8px 12px !important;
    font-family: 'Inter', sans-serif !important;
}

input:focus, select:focus {
    border-color: #38bdf8 !important;
    box-shadow: 0 0 0 2px rgba(56, 189, 248, 0.2) !important;
}

/* Animación fluida para los botones */
.custom-btn {
    background: linear-gradient(135deg, #0284c7 0%, #0369a1 100%) !important;
    color: white !important;
    font-weight: 600 !important;
    border-radius: 8px !important;
    border: none !important;
    padding: 10px 20px !important;
    cursor: pointer !important;
    transition: all 0.3s ease !important;
    box-shadow: 0 4px 6px -1px rgba(0, 0, 0, 0.2);
}

.custom-btn:hover {
    transform: translateY(-2px);
    box-shadow: 0 6px 12px -2px rgba(2, 132, 199, 0.4);
    background: linear-gradient(135deg, #0369a1 100%, #0284c7 0%) !important;
}

.custom-btn-success {
    background: linear-gradient(135deg, #059669 0%, #047857 100%) !important;
}

.custom-btn-success:hover {
    background: linear-gradient(135deg, #047857 0%, #059669 100%) !important;
    box-shadow: 0 6px 12px -2px rgba(5, 150, 105, 0.4);
}

/* Spinner de carga animado */
@keyframes pulse-glow {
    0% { opacity: 0.6; transform: scale(0.98); }
    50% { opacity: 1; transform: scale(1); text-shadow: 0 0 10px #38bdf8; }
    100% { opacity: 0.6; transform: scale(0.98); }
}

.loading-pulse {
    animation: pulse-glow 1.5s infinite ease-in-out;
    color: #38bdf8;
    font-weight: 600;
    font-size: 15px;
    padding: 15px;
    text-align: center;
}
</style>
""")

display(css_estilos)

# 4. COMPONENTES VISUALES AVANZADOS
input_api_key = widgets.Password(description="🔑 API Key:", placeholder="Gemini API Key", layout=widgets.Layout(width='450px'))
input_tema_macro = widgets.Text(description="🔍 Tema/País:", placeholder="Ej. Japón, Euro, Oro", layout=widgets.Layout(width='400px'))
input_ticker_macro = widgets.Text(description="📊 Ticker (Opc):", placeholder="Ej. EURUSD=X, GC=F", layout=widgets.Layout(width='400px'))

btn_analizar_macro = widgets.Button(description="🏛️ Generar Informe Macroeconómico", layout=widgets.Layout(width='320px', height='42px'))
btn_analizar_macro.class_name = 'custom-btn'

input_ticker_trading = widgets.Text(description="📊 Activo:", placeholder="Ej. EURUSD=X, BTC-USD, NVDA", layout=widgets.Layout(width='400px'))
select_temporalidad = widgets.Dropdown(
    options=['15m', '30m', '1h', '1d'],
    value='15m',
    description='⏱️ Temporal:',
    layout=widgets.Layout(width='300px')
)

btn_analizar_trading = widgets.Button(description="⚡ Analizar Estructura de Trading", layout=widgets.Layout(width='320px', height='42px'))
btn_analizar_trading.class_name = 'custom-btn custom-btn-success'

output_area = widgets.Output()

# Estructura de Pestañas
tab_macro = widgets.VBox([
    widgets.HTML("<p style='color: #94a3b8; font-size: 13px;'>Análisis profundo de economías, divisas y noticias geopolíticas globales.</p>"),
    input_tema_macro,
    input_ticker_macro,
    widgets.HTML("<br>"),
    btn_analizar_macro
], layout=widgets.Layout(padding='10px'))

tab_trading = widgets.VBox([
    widgets.HTML("<p style='color: #94a3b8; font-size: 13px;'>Análisis técnico de microestructura, niveles clave y RSI adaptado a tu temporalidad.</p>"),
    input_ticker_trading,
    select_temporalidad,
    widgets.HTML("<br>"),
    btn_analizar_trading
], layout=widgets.Layout(padding='10px'))

tabs = widgets.Tab(children=[tab_macro, tab_trading])
tabs.set_title(0, '🏛️ Macro & Divisas')
tabs.set_title(1, '⚡ Trading Táctico')

# Contenedor Visual Maestro
dashboard_html = widgets.HTML("""
<div class="fintech-container">
    <div class="fintech-header">
        <span>🌐</span> <span>Terminal Cuantitativa & Macroeconómica AI</span>
    </div>
</div>
""")

ui_maestro = widgets.VBox([
    dashboard_html,
    widgets.HTML("<br>"),
    input_api_key,
    widgets.HTML("<hr style='border-color: #334155;'>"),
    tabs,
    widgets.HTML("<hr style='border-color: #334155;'>"),
    output_area
])

# 5. LÓGICA DE EJECUCIÓN CON 30 REINTENTOS PARA ALTA DEMANDA
def ejecutar_macro(b):
    with output_area:
        clear_output()
        api_k = input_api_key.value.strip()
        tema = input_tema_macro.value.strip()
        ticker = input_ticker_macro.value.strip()

        if not api_k or not tema:
            display(HTML("<div style='color: #f87171; padding: 10px; background: rgba(248,113,113,0.1); border-radius: 8px;'>❌ Error: Debes ingresar tu API Key y el Tema o País a analizar.</div>"))
            return

        display(HTML("<div class='loading-pulse'>⏳ Conectando con fuentes globales y preparando inteligencia macro...</div>"))

        # Extracción de noticias
        busqueda_url = urllib.parse.quote(f"{tema} economía finanzas mercado")
        url = f"https://news.google.com/rss/search?q={busqueda_url}&hl=es-419&gl=US&ceid=US:es-419"
        try:
            feed = feedparser.parse(url)
            titulares = [f"- {e.title}" for e in feed.entries[:8]]
            noticias_str = "\n".join(titulares) if titulares else "Sin noticias recientes."
        except Exception as e:
            noticias_str = f"Error en noticias: {e}"

        # Extracción de datos de mercado
        datos_str = "Sin ticker provisto."
        if ticker:
            try:
                hist = yf.Ticker(ticker.upper()).history(period="1mo")
                if not hist.empty:
                    p_act = round(hist['Close'].iloc[-1], 2)
                    p_max = round(hist['High'].max(), 2)
                    p_min = round(hist['Low'].min(), 2)
                    tend = "ALCISTA 🟢" if p_act > hist['Close'].iloc[0] else "BAJISTA 🔴"
                    datos_str = f"Activo: {ticker.upper()} | Cierre: {p_act} | Máx: {p_max} | Mín: {p_min} | Tendencia: {tend}"
            except Exception as e:
                datos_str = f"Error al extraer ticker: {e}"

        prompt = f"""
        Actúa como un Estratega Macroeconómico Senior. Genera un informe riguroso basado en:
        - Tema: {tema}
        - Datos de Mercado: {datos_str}
        - Noticias Recientes:
        {noticias_str}

        Estructura con Markdown limpio y estos encabezados exactos:
        ### 📌 1. Resumen Ejecutivo (TL;DR)
        ### 📊 2. Entorno Macroeconómico y Cambiario
        ### 🌐 3. Factores Geopolíticos y Noticias Clave
        ### 🔮 4. Proyección de Escenarios
        """

        llamar_gemini_con_resiliencia(api_k, prompt, f"Macro_{tema.replace(' ', '_')}.txt")

def ejecutar_trading(b):
    with output_area:
        clear_output()
        api_k = input_api_key.value.strip()
        ticker = input_ticker_trading.value.strip().upper()
        temp = select_temporalidad.value

        if not api_k or not ticker:
            display(HTML("<div style='color: #f87171; padding: 10px; background: rgba(248,113,113,0.1); border-radius: 8px;'>❌ Error: Debes ingresar tu API Key y el Símbolo del Activo.</div>"))
            return

        display(HTML(f"<div class='loading-pulse'>⏳ Descargando velas intraday [{temp}] y calculando indicadores técnicos...</div>"))

        periodo_map = {"15m": "5d", "30m": "5d", "1h": "1mo", "1d": "3mo"}
        periodo = periodo_map.get(temp, "5d")

        try:
            hist = yf.Ticker(ticker).history(period=periodo, interval=temp)
            if hist.empty:
                display(HTML(f"<div style='color: #f87171;'>❌ No se encontraron datos para {ticker} en temporalidad {temp}.</div>"))
                return

            p_act = round(hist['Close'].iloc[-1], 2)
            resis = round(hist['High'].tail(15).max(), 2)
            sopor = round(hist['Low'].tail(15).min(), 2)

            delta = hist['Close'].diff()
            ganancia = (delta.where(delta > 0, 0)).rolling(14).mean()
            perdida = (-delta.where(delta < 0, 0)).rolling(14).mean()
            rs = ganancia / perdida
            rsi = round(float(100 - (100 / (1 + rs))).iloc[-1], 2) if not rs.empty else 50.0

            datos_tec = f"Activo: {ticker} | Temporalidad: {temp} | Precio Cierre: {p_act} | Resistencia: {resis} | Soporte: {sopor} | RSI(14): {rsi}"

        except Exception as e:
            display(HTML(f"<div style='color: #f87171;'>❌ Error al consultar Yahoo Finance: {e}</div>"))
            return

        prompt = f"""
        Actúa como un Trader Cuantitativo Senior experto en temporalidad de {temp}.
        Analiza esta estructura técnica:
        {datos_tec}

        Estructura la salida con Markdown y estos encabezados:
        ### ⚡ 1. Contexto de Precio y Momentum ({temp})
        ### 🛡️ 2. Niveles Operativos (Soportes y Resistencias)
        ### 📊 3. Lectura del RSI ({rsi})
        ### 🎯 4. Planteamiento Táctico y Gestión de Riesgo
        """

        llamar_gemini_con_resiliencia(api_k, prompt, f"Trading_{ticker}_{temp}.txt")

def llamar_gemini_con_resiliencia(api_key, prompt, nombre_archivo):
    client = genai.Client(api_key=api_key)
    modelo = 'gemini-3.8-flash'
    respuesta = None
    max_intentos = 30  # Sistema de hasta 30 intentos automáticos

    for intento in range(1, max_intentos + 1):
        try:
            respuesta = client.models.generate_content(
                model=modelo,
                contents=prompt,
                config=types.GenerateContentConfig(temperature=0.25)
            )
            if respuesta and hasattr(respuesta, 'text') and respuesta.text:
                break
        except Exception as e:
            error_str = str(e)
            if "503" in error_str or "unavailable" in error_str.lower():
                if intento < max_intentos:
                    time.sleep(3) # Espera inteligente antes del siguiente reintento
            else:
                clear_output()
                display(HTML(f"<div style='color: #f87171;'>❌ Error crítico de API: {error_str}</div>"))
                return

    clear_output()
    if respuesta and hasattr(respuesta, 'text') and respuesta.text:
        texto = respuesta.text
        with open(nombre_archivo, "w", encoding="utf-8") as f:
            f.write(texto)

        display(HTML(f"""
        <div class='fintech-container' style='border: 1px solid #059669;'>
            <div style='color: #34d399; font-weight: 600; margin-bottom: 10px;'>✨ ¡Análisis completado y guardado en '{nombre_archivo}'!</div>
        </div>
        <br>
        """))
        display(Markdown(texto))
    else:
        display(HTML("<div style='color: #f87171; padding: 10px; background: rgba(248,113,113,0.1); border-radius: 8px;'>🚨 Los servidores experimentaron alta demanda prolongada tras 30 intentos automáticos. Por favor, vuelve a hacer clic en el botón para reintentar.</div>"))

# Conectar botones
btn_analizar_macro.on_click(ejecutar_macro)
btn_analizar_trading.on_click(ejecutar_trading)

# Desplegar interfaz interactiva moderna
display(ui_maestro)
```
